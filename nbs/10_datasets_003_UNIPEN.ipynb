# UNIPEN Dataset

**Source**: [zenodo.org](https://zenodo.org/records/1195803)

UNIPEN is a (massive) database for on-line handwritten letters, digits and words. Recordings from over 2000 different writers with annotations are available. While the database is the result of a massive collaboration between research facilities, companies and private individuals, the data and associated websites and tools are partially outdated. This makes accessing parts of the data difficult. 

**Format**: Individual images of `28x28` pixels each. The dataset is provided in two formats. Images can be loaded using `scipy.io.loadmat function` or `.ToTensor()` transformation (see below).

A total of 6 splits is provided, see website for details.
